**Linear Regression using TensorFlowxt**

Objective: Build a simple linear regression model using only Tensors, Variables and GradientTape.

Suppose we have number of hours a student studies and their marks. Can we predict marks for a new student?

**Create the Dataset**

```python
import tensorflow as tf

x = tf.constant([1, 2, 3, 4, 5], dtype=tf.float32)
y = tf.constant([30, 40, 50, 60, 70], dtype=tf.float32)

print("X:", x)
print("Y:", y)
```

**Create Weights and Bias**

```python
w = tf.Variable(0.0)
b = tf.Variable(0.0)

print("Weight:", w)
print("Bias:", b)
```

These are our initial guesses

**Prediction**

```python
def predict(x):
  return w * x + b
```

```python
#Test
print(predict(x))
```

Initially:  
w = 0,  
b = 0

So prediction will be:  
[0, 0, 0, 0, 0]

But actual values are:  
[30, 40, 50, 60, 70]

Clearly our model is bad

**Calculate Error/Loss**

```python
# Mean Square Error(MSE)
def loss_function(y_true, y_pred):
  return tf.reduce_mean(tf.square(y_true - y_pred))    # return tf.reduce_mean((y_true-y_pred)**2)
```

```python
predictions = predict(x)

loss = loss_function(y, predictions)

print("Prediction:", predictions)
print("Loss:", loss)
```

Loss is huge because our predictions are terrible.

**Use GradientTape**

Using `GradientTape`, TensorFlow will change `w` and `b`

```python
with tf.GradientTape() as tape:
  predictions = predict(x)
  loss = loss_function(y,predictions)

gradients = tape.gradient(loss, [w, b])

print("Gradients:",gradients)

```

**Update the variables**

Gradient tells us how the loss changes. We use it to update our weight and bias.

```python
learning_rate = 0.01
```

```python
w.assign_sub(learning_rate * gradients[0])
b.assign_sub(learning_rate * gradients[1])
```

```python
print("Updated Weight:", w)
print("Updated Bias:", b)
```

we can see that the values have changed

A model learns by repeating this process many times

**Put it inside a Training Loop**

```python
learning_rate = 0.01
epochs = 1000

for epoch in range(epochs):

  with tf.GradientTape() as tape:
    predictions = predict(x)
    loss = loss_function(y, predictions)

  gradients = tape.gradient(loss, [w, b])

  w.assign_sub(learning_rate * gradients[0])
  w.assign_sub(learning_rate * gradients[1])

  if epoch % 100 == 0:
    print("Epoch", epoch, "Loss", loss.numpy())
```

At the beginning, our model's predictions are poor, so the loss is high. During training, GradientTape calculates the gradients and we update w and b. Therefore, the loss decreases. When the loss stops decreasing, it means the model has reached a point where further updates are not improving it significantly.

**See What the Model Learned**

```python
print("Final Weight:", w.numpy())
print("Final Bias:", b.numpy())
```

Model: y = 15x+1

which we learned from the data

**Make a Prediction**

```python
# what if student studies for 6 hours
new_x = tf.constant(6.0)

prediction = predict(new_x)

print("Predicted Marks:", prediction.numpy())
```
